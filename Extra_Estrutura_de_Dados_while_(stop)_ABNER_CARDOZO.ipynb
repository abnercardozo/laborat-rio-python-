# Exercício 1

Escrever um algoritmo que leia de forma enumerada os dados de 500 alunos de uma academia, a saber:

*   Altura ( $\mathrm{H > 0}$ ) em metros e
*   Massa ( $\mathrm{M > 0}$ ) em quilogramas


e exiba a tela seu IMC: Índice de Massa Corpórea, sabendo que:

<center>


IMC = $\dfrac{\strut \mathrm{M}}{\strut \mathrm{H ^ 2}}$

</center>

Exibir também na tela, em valores absolutos e relativos, os alunos por classificação:

*   SUBPESO - IMC: <= 18.5
*   SAUDÁVEL - IMC: ]18.5, 25.0]
*   SOBREPESO - IMC: > 25.0

Obrigatório:



*   Fazer o Tratamento de erro e exceção
*   Exibir todos os resultados formatados com uma (1) casa decimal de precisão.



```
try:
#ENTRADA(INPUT)
  altura = float(input("Insira a altura do aluno:"))
  massa = float(input("Informe o peso do aluno:"))

#PROCESSAMENTO (ULA)
  IMC = massa/altura**2

#SAÍDA(PRINT)
  if IMC <= 18.5:
    print("Aluno Subpeso")
  elif IMC > 18.5 and IMC < 25.0:
    print("Aluno Saudável")
  elif IMC > 25.0:
    print("Aluno Sobrepeso")

except(ValueError):
  print("Valor inválido")
```

```
num_alunos = 500
subpeso = 0
saudavel = 0
sobrepeso = 0

for i in range(1, num_alunos + 1):
    print(f"\n--- Aluno {i} ---")

    # ENTRADA (ALTURA)
    while True:
        try:
            altura = float(input("Insira a altura do aluno: "))
            if altura > 0:
                break
            else:
                print("Altura deve ser maior que 0")
        except ValueError:
            print("Valor inválido, digite um número")

    # ENTRADA (MASSA)
    while True:
        try:
            massa = float(input("Informe o peso do aluno: "))
            if massa > 0:
                break
            else:
                print("Massa deve ser maior que 0")
        except ValueError:
            print("Valor inválido, digite um número")

    # PROCESSAMENTO
    IMC = massa / altura ** 2

    if IMC <= 18.5:
        classificacao = "SUBPESO"
        subpeso += 1
    elif IMC <= 25.0:
        classificacao = "SAUDÁVEL"
        saudavel += 1
    else:
        classificacao = "SOBREPESO"
        sobrepeso += 1

    print(f"IMC: {IMC:.1f}")
    print(f"Classificação: {classificacao}")

#PRINT
print("\n===== RESULTADO GERAL =====")
print(f"Subpeso  : {subpeso} ({subpeso/num_alunos*100:.1f}%)")
print(f"Saudável : {saudavel} ({saudavel/num_alunos*100:.1f}%)")
print(f"Sobrepeso: {sobrepeso} ({sobrepeso/num_alunos*100:.1f}%)")
```

# Exercício 2

Escrever um algoritmo que leia de forma enumerada os dados de 1000 clientes de um e-commerce, a saber:

*   Idade (>= 18) em anos e
*   Salario bruto(>=  Salário Mínimo) em R$ 1621.01


e exiba na tela:
*   A média salarial dos clientes até 30 anos.
*   A média salarial dos clientes acima dos 30 anos.

```
clientes = 1000
soma_ate_30 = 0
contador_ate_30 = 0
soma_acima_30 = 0
contador_acima_30 = 0


for i in range (1, clientes +1):
  print(f"\n--- Cliente {i} ---")
  #ENTRADA (IDADE)
  while True:
    try:
      idade = float(input("Informe a idade do cliente:"))
      if idade >= 18:
        break
      else:
        print("Informe uma idade acima de 18.")

    except(ValueError):
      print("Valor inválido")

  #ENTRADA(SALARIO)
  while True:
    try:
      salario = float(input("Informe o salário bruto do cliente:"))
      if salario >= 1621.01:
          break
      else:
        print("Informe um salário acima do salário mínimo.")
    except(ValueError):
      print("Valor inválido")

  #PROCESSAMENTO
  if idade <= 30: #APLICA O CONTADOR
    soma_ate_30 += salario
    contador_ate_30 += 1
  else:
    soma_acima_30 += salario
    contador_acima_30 += 1


  #SAÍDA(PRINT)
  if contador_ate_30 > 0:
    media_ate_30 = soma_ate_30 / contador_ate_30
    print(f"\nMédia salarial dos clientes até 30 anos: R$ {media_ate_30:.2f}")
  else:
    print("\nNenhum cliente cadastrado até 30 anos.")

  if contador_acima_30 > 0:
    media_acima_30 = soma_acima_30 / contador_acima_30
    print(f"Média salarial dos clientes acima de 30 anos: R$ {media_acima_30:.2f}")
  else:
    print("Nenhum cliente cadastrado acima de 30 anos.")
```
