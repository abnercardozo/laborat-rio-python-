# QUESTÃO 2 (2.0 pontos): LISTA HOMOGÊNEA DE DADOS - CONTROLE DE DESPESAS

Desenvolva um programa **modular** (com funções) para controlar as despesas do mês (em R$) de um estudante. As despesas ficam em uma **lista homogênea**: todos os elementos são do tipo `float`.

1. **CRIAR / RESETAR A LISTA:** função que cria a lista vazia (e que também apaga todos os valores quando o usuário pedir um reset).
2. **MEMORIZAR A LISTA:** função que insere (memoriza) uma nova despesa no fim da lista.
3. **MODULARIZAÇÃO - DADO DE ENTRADA COM TRATAMENTO DE ERRO E EXCEÇÃO:** função que lê um número real **maior que zero**. Se o usuário digitar texto, deixar em branco ou digitar zero/negativo, exibir uma mensagem de erro e pedir novamente (`try / except` + `while`). Aceitar vírgula como separador decimal.
4. **MENU:** repetir até o usuário escolher sair:
   * 1 - Criar / Resetar a lista
   * 2 - Memorizar uma despesa
   * 3 - Exibir as despesas (numeradas, com 2 casas decimais)
   * 4 - Relatório: quantidade, total, média, maior e menor despesa
   * 5 - Pesquisar um valor na lista (informar se existe e em qual posição)
   * 0 - Sair

Opção de menu inválida (texto ou fora do intervalo) deve gerar mensagem de erro, sem derrubar o programa. Com a lista vazia, as opções 3, 4 e 5 devem avisar que não há despesas (nada de `ZeroDivisionError` nem `max()` de lista vazia).

**Obrigatório:** não usar `dict` nem `DataFrame`.

*(O formato desta questão foi inferido dos títulos das células do modelo da prova: CRIAR/RESETAR, MEMORIZAR, MODULARIZAÇÃO e MENU.)*

````python
# CRIAR / RESETAR A LISTA AQUI!
def criar_lista():
    return []          # lista homogênea: só números reais (float)

despesas = criar_lista()
print(despesas)
````

````python
# MEMORIZAR A LISTA AQUI!
def memorizar(lista, valor):
    lista.append(float(valor))

memorizar(despesas, 25.9)
print(despesas)
despesas = criar_lista()    # volta ao estado vazio antes do menu
````

````python
# MODULARIZAÇÃO: DADO DE ENTRADA COM TRATAMENTO DE ERRO E EXCEÇÃO
def ler_numero_positivo(mensagem):
    while True:
        try:
            valor = float(input(mensagem).replace(',', '.'))
            if valor <= 0:
                raise ValueError          # força o erro para zero ou negativo
            return valor
        except ValueError:
            print('Entrada inválida! Digite um número real maior que zero (ex.: 25.90).')

def ler_opcao(mensagem, minimo, maximo):
    while True:
        try:
            opcao = int(input(mensagem))
            if opcao < minimo or opcao > maximo:
                raise ValueError
            return opcao
        except ValueError:
            print(f'Opção inválida! Digite um número inteiro de {minimo} a {maximo}.')
````

**Atenção:** a célula do menu fica em um `while True`. Execute-a **separadamente** (não use "Executar tudo") e escolha 0 para sair.

````python
# CRIAR O MENU AQUI:
def exibir(lista):
    if len(lista) == 0:
        print('Não há despesas registradas.')
        return
    print(f'---{len(lista)} DESPESAS REGISTRADAS---')
    for indice, valor in enumerate(lista):
        print(f'{indice + 1}ª despesa: R$ {valor:.2f}')

def relatorio(lista):
    if len(lista) == 0:
        print('Não há despesas registradas.')
        return
    print('---RELATÓRIO DE DESPESAS---')
    print(f'Quantidade de despesas: {len(lista)}')
    print(f'Total gasto           : R$ {sum(lista):.2f}')
    print(f'Média das despesas    : R$ {sum(lista) / len(lista):.2f}')
    print(f'Maior despesa         : R$ {max(lista):.2f}')
    print(f'Menor despesa         : R$ {min(lista):.2f}')

def pesquisar(lista):
    if len(lista) == 0:
        print('Não há despesas registradas.')
        return
    valor = ler_numero_positivo('Valor a pesquisar (R$): ')
    if valor in lista:
        print(f'Valor encontrado! Posição: {lista.index(valor) + 1}ª despesa (ocorrências: {lista.count(valor)}).')
    else:
        print('Valor não encontrado na lista.')

def exibir_menu():
    print('\n=== CONTROLE DE DESPESAS ===')
    print('1 - Criar / Resetar a lista')
    print('2 - Memorizar uma despesa')
    print('3 - Exibir as despesas')
    print('4 - Relatório (quantidade, total, média, maior e menor)')
    print('5 - Pesquisar um valor')
    print('0 - Sair')

while True:
    exibir_menu()
    opcao = ler_opcao('Escolha uma opção: ', 0, 5)
    if opcao == 0:
        print('Programa encerrado.')
        break
    elif opcao == 1:
        despesas = criar_lista()
        print('Lista criada/resetada com sucesso.')
    elif opcao == 2:
        memorizar(despesas, ler_numero_positivo('Valor da despesa (R$): '))
        print('Despesa memorizada.')
    elif opcao == 3:
        exibir(despesas)
    elif opcao == 4:
        relatorio(despesas)
    elif opcao == 5:
        pesquisar(despesas)
````

**Pontos que a correção costuma olhar na Questão 2:**
* `try / except ValueError` dentro de um `while True`, com `return` só quando o dado é válido.
* Lista **vazia** tratada antes de `sum / len`, `max` e `min`.
* Cada operação em uma **função** (criar, memorizar, ler, exibir, relatório, pesquisar).
* Menu repetindo até a opção 0 e rejeitando opção inválida.
* Lista homogênea: só `float` (converta com `float(...)` ao memorizar).

Declare as constantes das colunas, as constantes de codificação e as listas de rótulos antes de resolver.

````python
import numpy as np

dataset = np.load('dataset.npy')

LINHAS = dataset.shape[0]
COLUNAS = dataset.shape[1]
print(f"LINHAS: {LINHAS} e COLUNAS: {COLUNAS}")

# Exibindo as 5 primeiras linhas somente para confirmar a carga (load) do dataset
dataset[:5]
````

````python
# CONSTANTES - COLUNAS
CODIGO = 0
SEXO = 1
PONTUACAO = 2
FATURA = 3
RELACIONAMENTO = 4

# CONSTANTES - CODIFICAÇÃO
FEMININO, MASCULINO = 1, 2
ESSENTIAL, VIP, PRIME = 0, 1, 2

# LISTAS HOMOGÊNEAS - RÓTULOS
rotulos_sexo = ["", "FEMININO", "MASCULINO"]
rotulos_relacionamento = ["ESSENTIAL", "VIP", "PRIME"]
````

## Letra A)

Uma campanha oferece **5% de cashback** sobre a última fatura dos sócios **VIP e PRIME** (os demais não recebem).

1. Crie, por vetorização, o vetor `cashback` (1000 posições; zero para quem não participa).
2. Use `enumerate` para exibir, em relatório enumerado, os **5 primeiros** sócios participantes: código, relacionamento, fatura e cashback.
3. Exiba o custo total da campanha.

````python
# 1) máscara dos participantes e vetor cashback (zero para os demais)
mascara_campanha = (dataset[:, RELACIONAMENTO] == VIP) | (dataset[:, RELACIONAMENTO] == PRIME)
cashback = np.zeros(LINHAS)
cashback[mascara_campanha] = 0.05 * dataset[mascara_campanha, FATURA]

# 2) relatório enumerado dos 5 primeiros participantes
socios_campanha = dataset[mascara_campanha]
cashback_campanha = cashback[mascara_campanha]

print(f'RELATÓRIO DA CAMPANHA: {len(socios_campanha)} sócios VIP/PRIME (5 primeiros):\n')
for indice, socio in enumerate(socios_campanha[:5]):
    print(f'Dados do {indice + 1}º Sócio:')
    print(f'Código(XXXX)      : {socio[CODIGO]:.0f}')
    print(f'Relacionamento    : {rotulos_relacionamento[int(socio[RELACIONAMENTO])]}')
    print(f'Última fatura (R$): {socio[FATURA]:.2f}')
    print(f'Cashback (R$)     : {cashback_campanha[indice]:.2f}\n')

# 3) custo total
print(f'Custo total da campanha: R$ {np.sum(cashback):,.2f}')
````

## Letra B)

Selecione os sócios **MASCULINOS** da categoria **ESSENTIAL** com **PONTUAÇÃO igual ou superior a 8000** e **última fatura inferior a R$ 1.000,00**.

3. Apresente a quantidade selecionada (absoluta e % do total de sócios) e, com `enumerate`, os 5 primeiros: ordem, código, pontuação e fatura formatada.

````python
mascara = (
    (dataset[:, SEXO] == MASCULINO)
    & (dataset[:, RELACIONAMENTO] == ESSENTIAL)
    & (dataset[:, PONTUACAO] >= 8000)
    & (dataset[:, FATURA] < 1000)
)
publico_alvo = dataset[mascara, :]

print(f'Quantidade selecionada: {len(publico_alvo)} ({len(publico_alvo) / LINHAS * 100:.1f}% dos sócios)\n')
for indice, socio in enumerate(publico_alvo[:5]):
    print(f'Dados do {indice + 1}º Sócio:')
    print(f'Código    : {socio[CODIGO]:.0f}')
    print(f'Pontuação : {socio[PONTUACAO]:.0f}')
    print(f'Fatura R$ : {socio[FATURA]:.2f}\n')
````

## Letra C)

Relatório das **5 sócias FEMININO com maior última fatura**, em ordem decrescente: CÓDIGO, FATURA e RELACIONAMENTO (rótulo).

*Dica:* primeiro filtre (máscara), depois ordene (`np.argsort`) e selecione com *fancy indexing*.

````python
socias = dataset[dataset[:, SEXO] == FEMININO]

indices_ordenados = np.argsort(socias[:, FATURA])      # crescente
indices_top5 = indices_ordenados[::-1][:5]             # decrescente, 5 primeiras
socias_top5 = socias[indices_top5]

print('RELATÓRIO DAS 5 SÓCIAS COM MAIOR FATURA:\n')
for indice, socia in enumerate(socias_top5):
    print(f'Dados da {indice + 1}ª sócia:')
    print(f'Código(XXXX)      : {socia[CODIGO]:.0f}')
    print(f'Última fatura (R$): {socia[FATURA]:.2f}')
    print(f'Relacionamento    : {rotulos_relacionamento[int(socia[RELACIONAMENTO])]}\n')
````

# COLA DE REVISÃO (padrões que mais se repetem)

| Quero... | Padrão |
|---|---|
| Condição composta em NumPy | `(a == X) & (b >= Y) \| (c < Z)` - parênteses em cada comparação |
| Quantidade / percentual | `np.sum(mascara)` · `quantidade / len(dataset) * 100` |
| Valores de um grupo | `dataset[mascara, COLUNA]` ou `vetor[mascara]` |
| Os N maiores | `np.argsort(v)[::-1][:N]` |
| Os N menores | `np.argsort(v)[:N]` |
| Posição do maior / menor | `np.argmax(v)` / `np.argmin(v)` (empate: primeira ocorrência) |
| Posições de uma condição | `np.argwhere(cond)[:, 0]` |
| Existe? depois localizar | `x in dataset[:, CODIGO]` + `np.argwhere(...)[0, 0]` |
| Média ponderada / composta | `(40 * a + 60 * b) / 100` - parênteses na soma |
| Lista heterogênea: filtrar | `[item for item in tabela if item[4] == 'SIM']` |
| Lista heterogênea: média | `sum(valores) / len(valores)` (cheque lista vazia) |
| Lista heterogênea: maior | `tabela[totais.index(max(totais))]` |
| Intervalo `[a, b[` | `a <= x < b` |
| Entrada segura | `while True` + `try / except ValueError` + `return` |
