```python
import numpy as np

numeros = np.array ([2,6,7,9,0,1,2,5,6,8])

print(numeros[-1])
print(numeros.dtype)

numeros[0] = 3
print(numeros[0])
```

# Array Bidimensional com Numpy


```python
M = np.array([[1,3,5], [2,4,6], [3,5,7]])
print(M) #CADA COLCHETE É UMA LINHA
```

```python
#PARA ACESSAR OS ELEMENTOS
print(M[0,1])
```

```python
M[0,1] = 7
print(M[0,1])
```

```python
print(M) #para verificar que houve a mudança dentro do array
```

```python
#VISUALIZAR UMA COLUNA INTEIRA
print('Primeira coluna:', M[:,0]) #USAMOS O SLICING PARA MOSTRAR A COLUNA
print('Segunda coluna', M[:,1])
```

```python
#VISUALIZAR LINHA INTEIRA
print('Segunda linha', M[1,:])
```

```python
#NÚMERO DE DIMENSÕES DE UM ARRAY:
print(numeros.ndim) #NDIM VERIFICA A DIMENSÃO DO ARRAY
print(M.ndim)
```

```python
#TAMANHO DO ARRAY
print(numeros.shape) #USAMOS O SHAPE PARA VERIFICAR O TAMANHO DO ARRAY
print(M.shape) #MATRIZ
```

```python
#NUMERO DE ELEMENTOS DENTRO DE UM ARRAY,
print(numeros.size) #PARA VERIFICAR OS ELEMENTOS
print(M.size)
```
