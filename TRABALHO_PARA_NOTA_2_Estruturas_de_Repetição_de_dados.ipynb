# EXERCÍCIO 3: EXERCÍCIO SORTEADO NÚMERO 9
Escrever um algoritmo que leia vários Números N (um por vez) que, no intervalo entre [10 90],
divididos por 5 possuem resto 2. Exiba a soma dos números lidos, parando o programa para N = 0.

```
soma = 0

while True:
  try:
    numeroN = int(input("Digite um número entre 10 e 90. Digite 0 para parar o programa."))
    if numeroN == 0:
      break
    elif numeroN < 10 or numeroN > 90:
      print("Número fora do intervalo")
    elif numeroN % 5 == 2:
      soma = numeroN + soma
  except ValueError:
    print("Digite um número inteiro")

print(f'A soma dos números é: {soma}') #ALINHADO AO WHILE
```

# EXERCÍCIO 4: EXERCÍCIO SORTEADO NÚMERO 13
Escrever um algoritmo, para cada item a seguir, que leia a altura em metros e o sexo (1: para Masculino
ou 2: para Feminino) de 50 pessoas. Este algoritmo deverá mostrar na tela:
a) A altura da pessoa mais alta por sexo.
b) Média da altura dos homens e das mulheres.
c) Quantidade de Homens e Mulheres, em porcentagem, com mais de 1.82 metros de altura.

```
entrevistados = 0
masculino = 0
feminino = 0
altura_masculino = 0  # contagem de homens > 1.82
altura_feminino = 0   # contagem de mulheres > 1.82
altura_mais_alta_masculino = 0
altura_mais_alta_feminino = 0
soma_altura_masculino = 0
soma_altura_feminino = 0

while entrevistados < 10:
    try:
        print('---MENU---')
        print('1: MASCULINO')
        print('2: FEMININO')
        print('3: SAIR')
        print('Escolha uma opção de 1 a 3')
        escolha = int(input())

        if escolha <= 0 or escolha >= 4:
            print("Escolha uma opção de 1 a 3")

        elif escolha == 3:
            break

        elif escolha == 1:
            altura = float(input('Digite a sua altura: '))
            masculino += 1
            entrevistados += 1
            soma_altura_masculino += altura
            if altura > altura_mais_alta_masculino:
                altura_mais_alta_masculino = altura
            if altura > 1.82:
                altura_masculino += 1

        elif escolha == 2:
            altura = float(input('Digite a sua altura: '))
            feminino += 1
            entrevistados += 1
            soma_altura_feminino += altura
            if altura > altura_mais_alta_feminino:
                altura_mais_alta_feminino = altura
            if altura > 1.82:
                altura_feminino += 1

    except ValueError:
        print("Valor inválido")

#PROCESSAMENTO FORA DO WHILE

media_altura_masculino = soma_altura_masculino / masculino if masculino > 0 else 0
media_altura_feminino = soma_altura_feminino / feminino if feminino > 0 else 0
palturamasculino = altura_masculino / masculino * 100 if masculino > 0 else 0
palturafeminino = altura_feminino / feminino * 100 if feminino > 0 else 0

print(f"A altura da pessoa mais alta masculina é {altura_mais_alta_masculino} metros e a altura da pessoa mais alta feminina é {altura_mais_alta_feminino} metros")
print(f"A média da altura dos homens é {media_altura_masculino:.2f} e a média da altura das mulheres é {media_altura_feminino:.2f}")
print(f"A porcentagem de homens com altura superior a 1.82 é {palturamasculino:.2f}% e de mulheres é {palturafeminino:.2f}%")
```

# EXERCÍCIO 5: EXERCÍCIO SORTEADO NÚMERO 16
Escreva um algoritmo que leia 300 números positivos e exiba o menor e o maior: par e ímpar.

```
contador = 0
menor_par = 0
maior_par = 0
menor_impar = 0
maior_impar = 0
primeiro_par = True
primeiro_impar = True

#ENTRADA(INPUT)
while contador < 300:
    try:
        numero = int(input(f"Digite o {contador + 1}º número: "))
        if numero <= 0:
            print("O número deve ser positivo.")
            continue
    except ValueError:
        print("Digite um número inteiro.")
        continue

    contador += 1

    if numero % 2 == 0:
        if primeiro_par:
            menor_par = numero
            maior_par = numero
            primeiro_par = False
        else:
            if numero < menor_par:
                menor_par = numero
            if numero > maior_par:
                maior_par = numero

    else:  # numero % 2 != 0
        if primeiro_impar:
            menor_impar = numero
            maior_impar = numero
            primeiro_impar = False
        else:
            if numero < menor_impar:
                menor_impar = numero
            if numero > maior_impar:
                maior_impar = numero

#SAÍDA
print(f"O menor número par é {menor_par} e o maior número par é {maior_par}")
print(f"O menor número ímpar é {menor_impar} e o maior número ímpar é {maior_impar}")
```

# EXERCÍCIO 8: EXERCÍCIO SORTEADO NÚMERO 25
Escreva um algoritmo que calcule a taxa média de glicose de cada um dos 500 pacientes em função de
sua idade, massa e se é diabético ou não, conforme a tabela:
image.png


```
import math
paciente = 0

while paciente < 3:
  try:
    idade = int(input("Informe sua idade:"))
    massa = int(input("Informe sua massa:"))
    diabete = int(input("Informe 1 se tiver diabete, 2 se não tiver diabetes"))

    if diabete == 1:
      taxa = math.sqrt(massa) / (0.93 * idade)
      paciente = paciente + 1
      print(f"A taxa média de glicose é: {taxa:.2f}")

    elif diabete == 2:
      taxa = math.sqrt(0.98 * massa) / (1.08 * idade)
      paciente = paciente + 1
      print(f"A taxa média de glicose é: {taxa:.2f}")

    else:
      print("Informe 1 se tiver diabete, 2 se não tiver diabetes")

  except(ValueError):
    print("Valor inválido")
```

# EXERCÍCIO 9: EXERCÍCIO SORTEADO NÚMERO 26
Escreva um algoritmo que leia o salário bruto de 10.000 entrevistados e exiba na tela a média salarial
entre as pessoas com escolaridade de ensino: Fundamental, Médio e Superior.

```
entrevistados = 0
fundamental = 0
medio = 0
superior = 0

soma_fundamental = 0
soma_medio = 0
soma_superior = 0

media_fundamental = 0
media_medio = 0
media_superior = 0

salario = 0


while entrevistados < 10000:
    try:
        print('---MENU---')
        print('ESCOLARIDADE')
        print('ENTREVISTADOS')
        print('1: FUNDAMENTAL')
        print('2: MÉDIO')
        print('3: SUPERIOR')
        print('4: SAIR')

        print('Escolha uma opção de 1 a 4')
        escolha = int(input())

        if escolha <= 0 or escolha >= 5:
            print("Escolha uma opção de 1 a 4")

        else:
            if escolha == 1:
                fundamental = fundamental + 1
                entrevistados = entrevistados + 1

                salario = int(input('Digite o salário: '))
                soma_fundamental = salario + soma_fundamental

            elif escolha == 2:
                medio = medio + 1
                entrevistados = entrevistados + 1

                salario = int(input('Digite o salário: '))
                soma_medio = salario + soma_medio

            elif escolha == 3:
                superior = superior + 1
                entrevistados = entrevistados + 1

                salario = int(input('Digite o salário: '))
                soma_superior = salario + soma_superior

            elif escolha == 4:
                break

    except ValueError:
        print("Valor inválido")

if fundamental > 0:
      media_fundamental = soma_fundamental / fundamental

if medio > 0:
      media_medio = soma_medio / medio

if superior > 0:
      media_superior = soma_superior / superior


print(f"A média salarial dos entrevistados com ensino fundamental completo é: {media_fundamental:.2f}")
print(f"A média salarial dos entrevistados com ensino médio completo é: {media_medio:.2f}")
print(f"A média salarial dos entrevistados com ensino superior completo é: {media_superior:.2f}")
```

# EXERCÍCIO 10: EXERCÍCIO SORTEADO NÚMERO 28
28) Faça um algoritmo que leia um valor inteiro positivo para a variável: ƞ (ƞ ≥ 1) e calcule e exiba na tela
o resultado da Soma (S) e da Multiplicação (M), a seguir:

```
import math

try:
    n = int(input("Digite um valor inteiro positivo para η: "))
except ValueError:
    print("Valor inválido")

soma = math.pi  # primeiro termo: π/1
multiplicacao = 1

for i in range(2, n + 1, 2):
    soma = soma + (math.pi / i)

for i in range(1, n + 1, 2):
    multiplicacao = multiplicacao * (i / math.pi)

print(f"S = {soma:.2f}")
print(f"M = {multiplicacao:.2f}")
```
