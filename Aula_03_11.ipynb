**USO DE IF E ELSE**

```python
faixa_de_preco_unitario = [
    [item[0], item[1], 'Até R$100' if item[2] <= 100 else 'Acima de R$100']
    for item in carrinho
]

for linha in faixa_de_preco_unitario:
  print(linha)
```

uso do sorted


```python
precos = [item [2] for item in carrinho]

precos_ordenados = sorted(precos)
precos_decrescente = sorted(precos, reverse = True)

print(f'Os preços ordenados são: {precos_ordenados}')
print(f'Preços descrescentes: {precos_decrescente}')
```

### Letra E:

Criar uma tabela nova: `novoCarrinho` inserindo a `coluna`: `IMPORTAÇÃO`, valor `SIM` ou `NÃO` e as colunas:

*   `CÓDIGO`
*   `PREÇO UNITÁRIO`
*   `QUANTIDADE`

Sabe-se que, as mercadorias do tipo `IMPORTAÇÃO` são as que:

*   Não tem `ISENÇÃO DE IMPOSTO` e
*   Tem o `PREÇO` superior a R$ 150.00

```python
# SOLUÇÃO PYTHÔNICA: por Lista Compactada
novoCarrinho = [
                [item[0], item[2], item[3], 'SIM' if (item[4] == 'NÃO' and item[2] > 150.00) else 'NÃO']
                for item in carrinho
               ]
novoCarrinho
```

```python
# SOLUÇÃO TRADICIONAL: Por laço de repetição e seleção (composta)
novoCarrinho = []
for item in carrinho:
  if(item[4] == 'NÃO' and item[2] > 150.00):
    novoCarrinho.append([item[0], item[2], item[3], 'SIM'])
  else:
    novoCarrinho.append([item[0], item[2], item[3], 'NÃO'])
novoCarrinho[:10]
```
