#LETRA G
Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no
mês. Calcule e mostre o total do seu salário no referido mês, sabendo-se que são descontados 11%
para o Imposto de Renda, 8% para o INSS e 5% para o sindicato. Com isso, exiba na tela:


*   salário bruto.
*   quanto pagou ao INSS.
*  quanto pagou ao sindicato.
*  o salário líquido = Brutos - Descontos.

````
try:
  #ENTRADA(INPUT)
  horasMes = float(input('Informa quantas horas você trabalha no mês:'))
  valorHora = float(input('Informa o quanto você ganha por hora:'))

  #PROCESSAMENTO
  salarioBruto = horasMes * valorHora
  impostoRenda = salarioBruto * 0.11
  inss = salarioBruto * 0.08
  sindicato = salarioBruto * 0.05
  salarioLiquido = salarioBruto - impostoRenda - inss - sindicato

  #SAÍDA(PRINT)
  print(f'O valor do salário bruto é: {salarioBruto:.2f}')
  print(f'O valor pago ao INSS é: {inss:.2f}')
  print(f'O valor pago ao sindicato é: {sindicato:.2f}')
  print(f'O valor do salário líquido é: {salarioLiquido:.2f}')

except(ValueError):
  print('Valor inválido')
````

#LETRA I
Calcular e exibir a distância entre dois pontos quaisquer do plano, P(x1, y1) e Q(x2, y2), sabendo que
a fórmula da distância é d = √(x2 – x1)

2 + (y2 – y1)

2, sendo os pontos P(x1, y1) e Q(x2, y2) como

dados de entrada.

````
try:
  #ENTRADA(INPUT)
  #IMPORTAÇÃO
  import math

  x1 = float(input('Informe o valor de x1:'))
  x2 = float(input('Informe o valor de x2:'))
  y1 = float(input('Informe o valor de y1:'))
  y2 = float(input('Informe o valor de y2:'))

  #PROCESSAMENTO
  distancia = math.sqrt((x2 - x1)**2 + (y2 - y1)** 2)

  #SAÍDA(PRINT)
  print(f'A distância entre os pontos é de: {distancia:.2f}')

except(ValueError):
  print('Valor inválido')
````

#LETRA J
Calcular e exibir o tempo (em horas) de autonomia de uma caixa d’água de um restaurante que
consome 1350 litros por hora em média. O tanque do restaurante é cilíndrico de base circular de Raio
(R) e de altura (H) em metros. Sabendo que 1 m3 = 1000 Litros.

````
try:
  #ENTRADA (INPUT)
  import math

  raio = float(input('Informe o valor do raio:'))
  altura = float(input('Informe o valor da altura:'))

  #PROCESSAMENTO
  volume = math.pi * raio**2 * altura
  litros = volume * 1000
  tempo = litros / 1350

  #SAÍDA(PRINT)
  print(f'O tempo de autonomia é de: {tempo:.2f} horas')

except(ValueError):
  print('Valor inválido')
````

velocidade de um link de Internet (em Megabytes / Segundo), calcule e informe o tempo: Minutos +
Segundos aproximado de download do arquivo usando este link.

````
try:
  #ENTRADA(INPUT)
  tamanhoArquivo = float(input('Informe o tamanho do arquivo em MegaByte:'))
  velocidadeInternet = float(input('Informe a velocidade da internet em Megabyte/Segundo:'))

  #PROCESSAMENTO
  tempo =  tamanhoArquivo / velocidadeInternet
  minutos = tempo // 60
  segundos = tempo % 60

  #SAÍDA(PRINT)
  print(f'O tempo de download é de: {tempo:.2f} minutos {minutos:.2f} e segundos {segundos:.2f}')

except(ValueError):
  print('Valor inválido')
````

#LETRA L
Calcular e exibir a distância máxima (em Quilômetros) de autonomia de um carro que possui um
tanque de combustível cúbico de lado (L) em metros e Altura (h) de preenchimento do tanque.
Sabendo que seu consumo é em média 10 km/litro. Sabendo que 1 m3 = 1000 Litros.

````
try:
  #ENTRADA(INPUT)
  import math

  lado = float(input('Informe o valor do lado:'))
  altura = float(input('Informe o valor da altura:'))

  #PROCESSAMENTO
  volume = math.pi * lado**2 * altura
  litros = volume * 1000
  mediaKm = litros / 10
  distancia = mediaKm * 1000

  #SAÍDA(PRINT)
  print(f'A distância máxima é de: {distancia:.2f} km')

except(ValueError):
  print('Valor inválido')
````
