```python
import pandas as pd
carrinho = pd.read_csv('carrinho.csv')
carrinho
```

```python
carrinho = carrinho.values.tolist()
carrinho[:5]
```

## Letra A:

Pesquisar uma mercadoria pelo código a escolha do usuário e exibir na tela todos os seus dados:

```python
#encontrar elemento pela indexação
while True: #para o usuário escolher o valor correto vamos utilizar o while
  #tratamento de exceção, lembrar disso para colocar na prova
  try:
    codigoProduto = int(input("Entre com o Código(XXX)"))
    #tratamento de erro:
    if(codigoProduto < 100 or codigoProduto > 999):
      print("ERRO NO VALOR DO CÓDIGO DO PRODUTO. Escolha novamente.")
    else:
      break
  except:
    print("ERRO DE EXCEÇÃO NO VALOR DO CÓDIGO DO PRODUTO. Escolha novamente.")
#LISTA COMPACTADA PARA FILTRAR A COLUNA CÓDIGO
listaCodigos = [item[0] for item in carrinho]
#usamos o [0] pois é onde o código está guardado
try:
  indiceProduto =  listaCodigos.index(codigoProduto)
  print(f'Produto localizado no {indiceProduto + 1} Item do carrinho:')
  print(f'Código: {carrinho[indiceProduto][0]}') #o [0] para não puxar todo o csv
  print(f'Nome do produto: {carrinho[indiceProduto][1]}')
  print(f'Preço: R${carrinho[indiceProduto][2]}')
  print(f'Quantidade: {carrinho[indiceProduto][3]}')
except:
  print(f'Código: {codigoProduto} não foi localizado no carrinho')
```
