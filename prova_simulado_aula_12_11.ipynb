```python
dataset = np.load('dataset.npy')

print("Estrutura:", type(dataset))
print("Tipo físico:", dataset.dtype)
LINHAS = dataset.shape[0]
COLUNAS = dataset.shape[1]
print(f"LINHAS: {LINHAS} X COLUNAS: {COLUNAS}")
```

```python
# CONSTANTES - COLUNAS:

CODIGO = 0
SEXO = 1
PONTUACAO = 2
FATURA = 3
RELACIONAMENTO = 4

# CONSTANTES - CODIFICAÇÃO:
FEMININO, MASCULINO = 1, 2
ESSENTIAL, VIP, PRIME = 0, 1, 2

print(f"Código do primeiro sócio: {dataset[0, CODIGO]: .0f}")
print(f"Última fatura: R$ {dataset[0, FATURA]:.2f}")

# LISTA HOMOGÊEA - RÓTULOS: Rótulos para apresentar os códigos das categorias nos relatórios.
rotulos_sexo = ["", "FEMININO", "MASCULINO"]
rotulos_relacionamento = ["ESSENTIAL", "VIP", "PRIME"]
```

```python
print(f"Código do 1º sócio: {dataset[0, CODIGO]: .0f}")
print(f"Última fatura do 1º sócio: R${dataset[0, FATURA]: .2f}")
print(f"Sexo do 1º sócio: {dataset[0, SEXO]: .2f}") #assim vai aparecer o valor do sexo, não o nome
```

```python
#código que aparece o rótulo correto do sexo para indicar o nome do sexo
print(f"Código do 1º sócio: {dataset[0, CODIGO]: .0f}")
print(f"Última fatura do 1º sócio: R${dataset[0, FATURA]: .2f}")
print(f"Sexo do 1° sócio              : {rotulos_sexo[int(dataset[0, SEXO])]}")
print(f"Tipo de relacionamento do 1º sócio: {rotulos_relacionamento[int(dataset[0, RELACIONAMENTO])]}") #nessa linha também forçamos o FLOAT de relacionamento virar INT
```
