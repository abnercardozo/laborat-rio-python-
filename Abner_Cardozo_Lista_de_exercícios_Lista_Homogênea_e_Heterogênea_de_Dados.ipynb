##Letra D:

Exiba um relatório administrativo com o Total da Folha de Pagamento:

*   Somatória: `SALÁRIO BRUTO`
*   Total Líquido = Somatória: `SALÁRIO BRUTO` - Descontos
*   Somatória: Descontos

 A empresa está descontando de quem tem `PLANO DE SAÚDE`:

*   R$ 212.15 dos funcionários `CLASSIFICAÇÃO`: OPERACIONAL

*   R$ 582.55 dos funcionários `CLASSIFICAÇÃO`: TÁTICO

*   R$ 992.01 dos funcionários `CLASSIFICAÇÃO`: ESTRATÉGICO


```
descontoOperacional = [item for item in empresa if item[2] == 'OPERACIONAL' and item [3] == 'SIM']
```

```
descontoTatico = [item for item in empresa if item[2] == 'TÁTICO' and item[3] == 'SIM']
```

```
descontoEstrategico = [item for item in empresa if item[2] == 'ESTRATÉGICO' and item[3] == 'SIM']
```

```
#SOMATORIO DO SALARIO BRUTO
salarioBruto = [item[4] for item in empresa]
```

```
#TOTAL LÍQUIDO
valorOP = 212.15
valorTA = 582.55
valorES = 992.01

descontoFinalOP = len(descontoOperacional)*valorOP
descontoFinalTA = len(descontoTatico)*valorTA
descontoFinalES = len(descontoEstrategico)*valorES

totalDescontos = descontoFinalOP + descontoFinalTA + descontoFinalES

totalLiquido = sum(salarioBruto) - totalDescontos
```

```
#RELATÓRIO
print(f'RELATÓRIO ADMINISTRATIVO')
print(f'TOTAL DA FOLHA DE PAGAMENTO')
print(f'SALÁRIO BRUTO: R$ {sum(salarioBruto):.2f}')
print(f'TOTAL LÍQUIDO: R$ {totalLiquido:.2f}')
print(f'DESCONTOS: R$ {totalDescontos:.2f}')
```

# Exercício 2:

Faça o carregamento do dataset: [triatlo.csv](https://drive.google.com/file/d/1QgJge2WOqdB7EF6eazbLYc4EN7dDLCHm/view?usp=sharing) e a partir da lista heterogênea desenvolva as seguintes soluções:

Triatletas diversos na transição costeira.png

```
import pandas as pd
triatlo = pd.read_csv('triatlo.csv')
triatlo
```

```
# LISTA HETEROGÊNEA:
triatlo = triatlo.values.tolist()
triatlo[:10]
```

##Letra A:

Exiba um relatório gerencial com a quantidade (em valores absolutos e relativos (%)) dos atletas com e sem `PATROCINADOR`.

```
comPatrocinador = [item for item in triatlo if item[1] != 'SEM PATROCÍNIO']
```

```
semPatrocinador = [item for item in triatlo if item[1] == 'SEM PATROCÍNIO']
```

```
#RELATÓRIO GERAL
print(f'---RELATÓRIO GERAL DOS ATLETAS DO TRIATLO---')
print(f'Quantidade de atletas sem patrocinador: {len(semPatrocinador)}')
print(f'Quantidade de atletas com patrocinador: {len(comPatrocinador)}')

print(f'---RELATÓRIO PERCENTUAL---')
print(f'Porcentam de atletas com patrocinador: {len(comPatrocinador)/len(triatlo)*100:.2f}%')
print(f'Porcentagem de atletas sem patrocinador: {len(semPatrocinador)/len(triatlo)*100:.2f}%')
```

##Letra B:

Exiba um relatório com a `MATRÍCULA`, `PATROCINADOR` e tempo total = `NATAÇÃO (min)` + `CORRIDA (min)`+ `CICLISMO (min)`.

Ou seja, este relatório deve exibir esses dados do vencedor:
*   `NATAÇÃO`
*   `CORRIDA`
*   `CICLISMO`
* GERAL = `NATAÇÃO (min)` + `CORRIDA (min)`+ `CICLISMO (min)`

```
tempoTotal = [item[5] + item[6] + item[7] for item in triatlo]
```

```
atletaVencedor = [item for item in triatlo if min(tempoTotal) == item[5] + item[6] + item[7]]
```

```
#RELATÓRIO DO VENCEDOR
print(f'---DADOS DO VENCEDOR---')
for item in atletaVencedor:
  print(f'MATRÍCULA:    {item[0]}')
  print(f'PATROCINADOR: {item[1]}')
  print(f'TEMPO DE NATAÇÃO: {item[5]:.2f} minutos')
  print(f'TEMPO DE CORRIDA: {item[6]:.2f}  minutos')
  print(f'TEMPO DE CICLISMO: {item[7]:.2f}  minutos')
  print(f'TEMPO TOTAL DO VENCEDOR: {item[5] + item[6] + item[7]} minutos')
```

```
#RELATÓRIO DO VENCEDOR SOLUÇÃO COM INDEX
print(f'---DADOS DO VENCEDOR---')
print(f'Matrícula: {triatlo[tempoTotal.index(min(tempoTotal))][0]}')
print(f'Patrocinador: {triatlo[tempoTotal.index(min(tempoTotal))][1]}')
print(f'TEMPO DE NATAÇÃO: {triatlo[tempoTotal.index(min(tempoTotal))][5]:.2f}')
print(f'TEMPO DE CORRIDA: {triatlo[tempoTotal.index(min(tempoTotal))][6]:.2f}')
print(f'TEMPO DE CICLISMO: {triatlo[tempoTotal.index(min(tempoTotal))][7]:.2f}')
print(f'TEMPO TOTAL DO VENCEDOR: {min(tempoTotal):.2f} minutos')
```

##Letra C:

Exiba um relatório gerencial com a quantidade (em valores absolutos e relativos (%)) de atletas:
*   Sedentários: IMC = `MASSA` / `ALTURA` $^ 2$ entre [18.5, 25.0[.
*   Caso contrário.

```
atletaSedentario = [item for item in triatlo if item[3]/item[4]**2 >= 18.5 and item[3]/item[4]**2 <= 25.0]
print(f'{len(atletaSedentario)}')
```

```
atletaSaudavel = len(triatlo) - len(atletaSedentario)
```

```
print(f'---RELATÓRIO GERAL---')
print(f'QUANTIDADE DE ATLETAS SEDENTÁRIOS: {len(atletaSedentario)}')
print(f'PORCENTAGEM DE ATLETAS SEDENTÁRIOS: {len(atletaSedentario)/len(triatlo)*100:.2f}%')
print(f'QUANTIDADE DE ATLETAS NÃO SEDENTÁRIOS: {(atletaSaudavel)}')
print(f'PORCENTAGEM DE ATLETAS NÃO SEDENTÁRIOS: {(atletaSaudavel)/len(triatlo)*100:.2f}%')
```

##Letra D:

Exiba um relatório com a `MATRÍCULA`, `IDADE` dos atletas com mais de 40 anos que fizeram um tempo abaixo do tempo médio geral do triatlo considerando as 3 modadlidades = `NATAÇÃO (min)` + `CORRIDA (min)`+ `CICLISMO (min)`.

```
atletaQuarenta = [item for item in triatlo if item[2] >= 40]
```

```
tempoMedio = sum(tempoTotal)/len(tempoTotal)
```

```
#RELATÓRIO GERAL

print(f'---RELATÓRIO GERAL---')
for indice, item in enumerate(atletaQuarenta):
  print(f'DADOS DO {indice + 1}º ITEM: ')
  print(f'DADOS DOS {len(atletaQuarenta)} ATLETAS ACIMA DE 40 ANOS')
  print(f'MATRÍCULA: {item[0]}')
  print(f'IDADE: {item[2]}')
  print(f'TEMPO MÉDIO: {tempoMedio:.2f} minutos')
```

# Exercício 3:

Faça o carregamento do dataset: [clube_de_vantagens.csv](https://drive.google.com/file/d/1d5gO2x77TP7h5d7JLPTKoUzK3TF8P0Im/view?usp=sharing) e a partir da lista heterogênea desenvolva as seguintes soluções:

Encontro do clube de benefícios.png

```
import pandas as pd
clube_de_vantagens = pd.read_csv('clube_de_vantagens.csv')
clube_de_vantagens
```

```
# LISTA HETEROGÊNEA:
clube_de_vantagens = clube_de_vantagens.values.tolist()
clube_de_vantagens[:10]
```

##Letra A:

Exiba um relatório gerencial com a quantidade (em valores absolutos e relativos (%)) dos sócios por `RELACIONAMENTO`:

```
socioEssential = [item for item in clube_de_vantagens if item[2] == 'ESSENTIAL']
```

```
socioVip = [item for item in clube_de_vantagens if item[2] == 'VIP']
```

```
socioPrime = [item for item in clube_de_vantagens if item[2] == 'PRIME']
```

```
print(f'---RELATÓRIO SÓCIOS---')
print(f'A quantidade de sócios na categoria ESSENTIAL é de {len(socioEssential)}')
print(f'A porcentagem de sócios na categoria ESSENTIAL é de {len(socioEssential)/len(clube_de_vantagens)*100:.2f}%')
print(f'A quantidade de sócios na categoria VIP é de {len(socioVip)}')
print(f'A porcentagem de sócios na categoria VIP é de {len(socioVip)/len(clube_de_vantagens)*100:.2f}%')
print(f'A quantidade de sócios na categoria PRIME é de {len(socioPrime)}')
print(f'A porcentagem de sócios na categoria PRIME é de {len(socioPrime)/len(clube_de_vantagens)*100:.2f}%')
```

##Letra B:

Exiba um relatório gerencial com o gasto médio da `FATURA - CARTÃO DE CRÉDITO (R$)` dos sócios por `RELACIONAMENTO`:

```
gastoMedioE = [item[5] for item in clube_de_vantagens if item[2] == 'ESSENTIAL']
```

```
gastoMedioV = [item[5] for item in clube_de_vantagens if item[2] == 'VIP']
```

```
gastoMedioP = [item[5] for item in clube_de_vantagens if item[2] == 'PRIME']
```

```
#RELATÓRIO GERAL
print(f'---RELATÓRIO GERAL---')
print(f'O gasto médio dos sócios da categoria ESSENTIAL é de R$ {sum(gastoMedioE)/len(gastoMedioE):.2f}')
print(f'O gasto médio dos sócios da categoria VIP é de R$ {sum(gastoMedioV)/len(gastoMedioV):.2f}')
print(f'O gasto médio dos sócios da categoria PRIME é de R$ {sum(gastoMedioP)/len(gastoMedioP):.2f}')
```

##Letra C:

Exiba um relatório gerencial com a quantidade (em valores absolutos e relativos (%)) de sócios por `RELACIONAMENTO` que gastam acima do gasto médio:

##Letra D:

Exiba um relatório gerencial, separado por `SEXO`, com os dados dos sócios: `NOME` e `FATURA - CARTÃO DE CRÉDITO (R$)` que tem uma `PONTUAÇÃO` acima de 95% da `PONTUAÇÃO` máxima entre todos os sócios:

```
pontuacaoMax = max([item[4] for item in clube_de_vantagens])
limite = pontuacaoMax * 0.95
```

```
sexoM = [item for item in clube_de_vantagens if item[3] == 'MASCULINO' and item[4] > limite]
```

```
sexoF =[item for item in clube_de_vantagens if item[3] == 'FEMININO' and item [4] > limite]
```

```
#RELATORIO GERAL

print(f'---RELATÓRIO DE {len(sexoM)} SÓCIOS MASCULINOS ACIMA DE 95%---')
for item in sexoM:
  print(f'NOME: {item[1]}')
  print(f'FATURA: R$ {item[5]:.2f}')

print(f'---RELATÓRIO DE {len(sexoF)} SÓCIOS FEMININOS ACIMA DE 95%---')
for item in sexoF:
  print(f'NOME: {item[1]}')
  print(f'FATURA: R$ {item[5]:.2f}')
```
