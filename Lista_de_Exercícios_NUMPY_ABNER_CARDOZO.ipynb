## Letra B)

Exibir na Tela: Média de tempo de acesso por clientes dos quartis (Q1 = 25%), (Q2 = 50%) e (Q3 = 75%)) do dataset.

```python
q1 = np.percentile(dataset6[:, TEMPO_MENSAL], 25)
q2 = np.percentile(dataset6[:, TEMPO_MENSAL], 50)
q3 = np.percentile(dataset6[:, TEMPO_MENSAL], 75)
```

```python
mascara_1 = (dataset6[:, TEMPO_MENSAL] < q1)
mascara_2 = (dataset6[:, TEMPO_MENSAL] >= q1) & (dataset6[:, TEMPO_MENSAL] < q2)
mascara_3 = (dataset6[:, TEMPO_MENSAL] >= q1) & (dataset6[:, TEMPO_MENSAL] >= q2) & (dataset6[:, TEMPO_MENSAL] < q3)
```

```python
media_1 = np.mean(dataset6[mascara_1, TEMPO_MENSAL])
media_2 = np.mean(dataset6[mascara_2, TEMPO_MENSAL])
media_3 = np.mean(dataset6[mascara_3, TEMPO_MENSAL])

print(f"---RELATÓRIOS DA MÉDIA DE TEMPO DE ACESSO DOS CLIENTES---")
print(f"Média do tempo de acesso de clientes em Q1: {media_1:.2f} ")
print(f"Média do tempo de acesso de clientes em Q2: {media_2:.2f} ")
print(f"Média do tempo de acesso de clientes em Q3: {media_3:.2f} ")
```
