*   True: Número é primo
*   False: Número não é primo
Regras de teste de um número primo:
*   Um número primo precisa ter exatamente dois divisores diferentes: O numero 1 e ele mesmo. Ou seja, 2 e 3 são números primos
*   O número 1 não é um número primo
*   Um número qualquer não pode conter nenhum múltiplo no intervalo fechado entre [2, numero // 2 +1]

Na célula do programa principal, use a função várias vezes, criando um Menu para o usuário, conforme as opções a seguir:


*   [1]: Exibir uma lista enumerada dos núumeros primos entre [1, número]
*   [QUALQUER TECLA]: Sair do programa










```python
#CRIAR A FUNÇÃO AQUI
def soma(STOP):
  S1, S2 = 0, 0
  for contador in range(100, STOP + 1):
    if (contador % 7 == 0):
      if (contador % 2 == 0):
        S1 = S1 + contador
      else:
        S2 = S2 + contador
  return S1, S2
```

```python
#MODULARIZAÇÃO DE DADOS DE ENTRADA
def lerSTOP():
  while True:
    try:
      STOP = int(input("Escolha um número inteiro de [100, 999] "))
      if(STOP < 100 or STOP > 999):
        print("ERRO NO VALOR DO STOP: Escolha novamente")
      else:
        return STOP
    except:
      print("ERRO DE EXCEÇÃO NO VALOR DO STOP: Escolha novamente")
```
