# EXERCÍCIO SORTEADO 4
Tendo como dado de entrada a altura (h) e o sexo de uma pessoa, construa um algoritmo que calcule
seu peso (Massa: Quilogramas) ideal, utilizando as seguintes fórmulas:

✓ Para homens: (72.7 ∗ h) − 58
✓ Para mulheres: (62.1 ∗ h) − 44.7

```
# ALGORITMO: ENTRADA (Usuário) + PROCESSAMENTO (ULA) + SAÍDA (Objetivo)
try:
#ENTRADA(INPUT)
  sexo = input('Informe o sexo(homem/mulher):')
  altura = float(input('Informe a altura:'))

#PROCESSAMENTO (ULA)
  if sexo == 'homem':
     massa = (72.7 * altura) - 58
  elif sexo == 'mulher':
    massa = (62.1 * altura) - 44.7
  else:
    print('Sexo inválido')

#SAÍDA(PRINT)
  print(f'O peso ideal é: {massa:.2f} kg')

except(ValueError):
  print('Valor inválido')
```

# EXERCÍCIO SORTEADO 13
Escrever um algoritmo em C que leia três (3) números reais quaisquer e exiba o cubo da média deles
se está média estiver fora do intervalo fechado [10Φ, 200Φ]. Caso contrário, exiba a própria média.
Valor de Φ (PHI = 11.52743: Use 5 casas decimais: ":5"

```
try:
  #ENTRADA(INPUT)
  num1 = float(input('Informe o primeiro número:'))
  num2 = float(input('Informe o segundo número:'))
  num3 = float(input('Informe o terceiro número:'))

  #PROCESSAMENTO(ULA)
  PHI = 11.52743
  media = (num1 + num2 + num3)/3

  if media < 10*PHI or media > 200*PHI:
    mediacubo = media**3
    print(f'A média está fora do intervalo: {mediacubo:.5f}')
  else:
    print(f'A média está dentro do intervalo: {media:.5f}')

except(ValueError):
  print('Valor inválido')
```

# EXERCÍCIO SORTEADO 14

da UVV, sendo que:
image.png
E; exiba na tela seus Resultados: Parciais e Final (STATUS: Aprovado, Prova Final, Reprovado por Falta ou Reprovado).

```
try:

    # ENTRADA (INPUT)
    av1 = float(input('Insira a nota da 1ª avaliação: '))
    av2 = float(input('Insira a nota da 2ª avaliação: '))
    tf = float(input('Insira o total de faltas: '))

    # PROCESSAMENTO (ULA)
    mediaparcial = (av1 + av2) / 2

    if tf > 20:
        print('Aluno Reprovado por Falta')

    elif mediaparcial >= 7:
        print('Aprovado')

    else:
        print('Prova Final')

except ValueError:
    print('Valor inválido')
```

# EXERCÍCIO SORTEADO 16

se os valores informados forem iguais) e exiba na saída a média dos dois maiores números lidos.
