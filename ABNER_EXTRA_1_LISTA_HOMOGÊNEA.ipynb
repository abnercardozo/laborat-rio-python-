## Questão 1: Exibir a Lista Enumerada em Ordem:

*   Alfabética (invertida):
*   Pelo tamanho da string:


```
municipios_ordenados = municipios.copy()
municipios_ordenados.sort(reverse=True)

for i in range (len(municipios_ordenados)):
  print(f'{i+1}º - {municipios_ordenados[i]}')
```

```
municipios_tamanho = municipios.copy()
municipios_tamanho.sort(key=len)

for i in range(len(municipios_tamanho)):
    print(f'{i + 1} - {municipios_tamanho[i]} ({len(municipios_tamanho[i])} letras)')
```

## Questão 2: Exibir os dados completo do município com mais e menos letras no nome:

```
maior_nome = max(municipios, key=len)
menor_nome = min(municipios, key=len)

print('MUNICÍPIO COM MAIS LETRAS')
print('Nome:', maior_nome)
print('População:', populacao[municipios.index(maior_nome)])
print('Área:', area[municipios.index(maior_nome)])

print()

print('MUNICÍPIO COM MENOS LETRAS')
print('Nome:', menor_nome)
print('População:', populacao[municipios.index(menor_nome)])
print('Área:', area[municipios.index(menor_nome)])
```

## Questão 3: Pesquisar o nome de um município a escolha do usuário (input).

```
busca = input("Digite o nome de um município:")

if busca in municipios:
  posicao = municipios.index(busca)

  print("Município", municipios[posicao])
  print("População:", populacao[posicao])
  print("Área:", area[posicao])
else:
  print("Município não encontrado")
```

## Questão 4: Exibir uma lista enumerada dos municípios com a sua respectiva densidade populacional:

Densidade Populacional = População / Área

```
print('DENSIDADE POPULACIONAL DE CADA MUNICPIO')

for i in range (len(municipios)):
  densidade = populacao[i] / area[i]
  print(f'{i+1} - {municipios[i]} ({densidade:.2f})')
```

## Questão 5: Pesquisar as regiões do E.S e exibir a população total de cada região:

```
regiao_metropolitana = [
    'Vitória',
    'Vila Velha',
    'Serra',
    'Cariacica',
    'Viana',
    'Guarapari',
    'Fundão'
]

regiao_sul = [
    'Alfredo Chaves',
    'Anchieta',
    'Iconha',
    'Piúma',
    'Itapemirim',
    'Marataízes',
    'Rio Novo do Sul',
    'Cachoeiro de Itapemirim',
    'Muqui',
    'Atílio Vivácqua',
    'Apiacá',
    'Mimoso do Sul',
    'Presidente Kennedy',
    'Bom Jesus do Norte',
    'Vargem Alta'
]

regiao_norte = [
    'Jaguaré',
    'São Mateus',
    'Conceição da Barra',
    'Vila Valério',
    'São Domingos do Norte',
    'São Gabriel da Palha',
    'Nova Venécia',
    'Boa Esperança',
    'Pinheiros',
    'Pedro Canário',
    'Montanha',
    'Mucurici',
    'Ponto Belo',
    'Vila Pavão',
    'Águia Branca',
    'Barra de São Francisco',
    'Ecoporanga',
    'Água Doce do Norte',
    'Mantenópolis',
    'Alto Rio Novo'
]
```

```
total_norte = 0
total_sul = 0
total_metropolitana = 0

for i in range(len(municipios)):

    if municipios[i] in regiao_norte:
        total_norte += populacao[i]

    elif municipios[i] in regiao_sul:
        total_sul += populacao[i]

    elif municipios[i] in regiao_metropolitana:
        total_metropolitana += populacao[i]

print('Região Norte:', total_norte)
print('Região Sul:', total_sul)
print('Região Metropolitana:', total_metropolitana)
```
