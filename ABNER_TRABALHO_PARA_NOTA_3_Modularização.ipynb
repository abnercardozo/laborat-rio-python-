# EXERCÍCIO 1: EXERCÍCIO SORTEADO NÚMERO 4
Criar uma função e seu protótipo que receba o valor de um ano e retorne da função:
Zero (0): Ano NÃO é Bissexto ou Um (1): Ano é Bissexto. Regras do ano Bissexto (Ver exercício 11):
Na célula do programa principal, use a função 100 vezes para valores de anos lidos do usuário (um de
cada vez).

```
# CRIAR A FUNÇÃO AQUI
def ano_bissexto(ano):
  if ano % 400 == 0 or (ano % 4 == 0 and ano % 100 != 0):
    return 1
  else:
      return 0

```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_ano():
  while True:
    try:
      ano = int(input("Digite o ano:"))
      #TRATAMENTO DE ERRO
      if (ano <= 0):
        print("Digite o ano novamente.")
      else:
        return ano
    except:
      print("Erro na digitação. Escolha o ano novamente.")
```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
contagem_ano = 0

print("1: Ano Bissexto")
print("0: Ano não é Bissexto")

while True:
    ano = ler_ano()
    if ano_bissexto(ano) == 1:
      print(f"1: Ano {ano} Bissexto")
    else:
      print(f"0: Ano {ano} não é Bissexto")
    contagem_ano += 1
    print(f"Quantidade de anos analisados: {contagem_ano}")
    if contagem_ano == 100:
        break
```

# EXERCÍCIO 2: EXERCÍCIO SORTEADO NÚMERO 6
Crie uma função que calcule o IMC (Índice de Massa Corpórea) de uma pessoa com base na sua idade,
massa e sexo conforme tabela. Na célula do programa principal, use a função e exiba o IMC de 100
entrevistados (um por um).
image.png

```
# CRIAR A FUNÇÃO AQUI
def calcular_IMC(massa, altura, sexo):
  if sexo == 1:
    IMC_FEMININO = 0.95 * massa / (altura * altura)
    return IMC_FEMININO
  elif sexo == 2:
    IMC_MASCULINO = 1.05 * massa / (altura * altura)
    return IMC_MASCULINO
```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_entrevistado():
  while True:
    #TRATAMENTO DE EXCEÇÃO
    try:
      print("MENU")
      print("1: Feminino")
      print("2: Masculino")

      sexo = int(input("Digite o número para seu sexo:"))
      if sexo == 1 or sexo == 2:
        massa = float(input("Digite sua massa"))
        altura = float(input("Digite sua altura:"))

      return sexo, massa, altura

    except ValueError:
      print("Digite um valor válido")

```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
entrevistados = 0

while entrevistados < 100:
  entrevistados += 1
  sexo, massa, altura = ler_entrevistado()
  imc = calcular_IMC(massa, altura, sexo)
  print(f"O IMC do entrevistado {entrevistados} é {imc:.2f}")
```

# EXERCÍCIO 3: EXERCÍCIO SORTEADO NÚMERO 8
Faça um programa Python modularizado que calcule o somatório dos n primeiros termos da série de
Fibonacci ( 0, 1, 1, 2, 3, 5, 8, ...). A função deve receber o valor de n e retornar o valor do somatório
para ser impresso na célula do programa principal.

```
# CRIAR A FUNÇÃO AQUI
def somatoria(n):
  a, b = 0, 1
  soma = 0
  for i in range(n):
    soma += a
    a, b = b, a + b
  return soma
```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_termo():
  while True:
    try:
      n = int(input("Digite o valor de n:"))
      if n <= 0: #TRATAMENTO DE ERRRO
        print("Digite um valor válido")
      else:
        return n
    except ValueError:
      print("Digite um valor válido")
```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
n = ler_termo()
resultado = somatoria(n)
print(f"O somatório dos {n} primeiros termos da série de Fibonacci é {resultado}")
```

# EXERCÍCIO 4: EXERCÍCIO SORTEADO NÚMERO 11
 Crie uma função e seu protótipo que receba como argumento um ano qualquer e um dos três números
citados: 4 ou 100 ou 400 e retorne zero ou um (se o ano for ou não bissexto), quociente e resto da
divisão do ano pelo número escolhido, conforme as regras de teste do ano bissexto. Após isso, use a
função no programa principal e exiba as informações sobre o ano conforme os dados ditos pelo usuário.
Regras de teste do ano bissexto:
✔ São bissextos todos os anos múltiplos de 400. Exemplo: 1600, 2000, 2400,2800...
✔ São bissextos todos os múltiplos de 4 e não múltiplos de 100. Exemplo: 1996, 2004, 2008, ...
✔ Não são bissextos todos os demais anos.

```
# CRIAR A FUNÇÃO AQUI
def ano_bissexto(divisor, ano):
  if ano % 400 == 0 or (ano % 4 == 0 and ano % 100 != 0):
    bissexto = 1
  else:
      bissexto = 0

  quociente = ano // divisor
  resto = ano % divisor
  return bissexto, quociente, resto
```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_ano():
  while True:
    try:
      ano = int(input("Digite um ano:"))
      if ano <= 0:
        print("Digite um ano válido.")
      else:
        return ano
    except ValueError:
      print("Digite um valor válido")

def ler_divisor():
  while True:
    try:
      divisor = int(input("Escolha um divisor entre 4, 100 ou 400:"))
      if divisor in (4, 100, 400):
        return divisor
      else:
        print("Digite um divisor válido.")
    except ValueError:
      print("Digite um valor válido")
```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
ano = ler_ano()
divisor = ler_divisor()
bissexto, quociente, resto = ano_bissexto(divisor, ano)

if bissexto == 1:
  print(f"O ano {ano} é bissexto")
else:
  print(f"O ano {ano} não é bissexto")

print(f"O ano {ano} tem como quociente o valor {quociente} e como resto o valor {resto}")
```

# EXERCÍCIO 5: EXERCÍCIO SORTEADO NÚMERO 12.
Crie uma função e seu protótipo que receba como argumento um número inteiro qualquer
representando os segundos de realização de um teste de laboratório e retorne as horas, minutos e
segundos convertidos deste número. Use a função (100 vezes) no programa principal e exiba a
informação, conforme o exemplo a seguir:
Exemplo: Tempo: 10000 Segundos = 2 Horas(s) + 46 Minuto(s) + 40 Segundos(s).

```
# CRIAR A FUNÇÃO AQUI
def converter_tempo(segundos):
  horas = segundos // 3600
  segundos_restantes = segundos % 3600
  minutos = segundos_restantes // 60
  segundos_restantes = segundos_restantes % 60
  return horas, minutos, segundos_restantes
```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_segundos():
  while True:
    try:
      segundos = int(input("Digite o número de segundos:"))
      if segundos <= 0:
        print("Digite segundos acima de zero.")
      else:
        return segundos
        contagem += 1
    except ValueError:
      print("Digite um valor válido")
```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
contagem = 0

while contagem < 100:
  segundos = ler_segundos()
  horas, minutos, segundos_restantes = converter_tempo(segundos)
  contagem += 1
  print(f"O tempo {segundos} segundos da prova equivale a {horas} horas, {minutos} minutos, {segundos} segundos e {segundos_restantes} segundos restantes")
```

# EXERCÍCIO 7: EXERCÍCIO SORTEADO NÚMERO 14
Crie uma função e seu protótipo que receba a hora de entrada e saída de um estacionamento e retorne o
tempo de permanência e total a pagar. Na célula do programa principal o usuário informa:
Hora de entrada (HH:MM): Exemplo: 13:37
Hora de saída (HH:MM): Exemplo: 18:41 (Menus)
Diferença: (HH:MM): Exemplo: 04:64 → 5 Horas de permanência no estacionamento.

```
# CRIAR A FUNÇÃO AQUI
def diferenca_estacionamento(hora_entrada, minuto_entrada, hora_saida, minuto_saida):
  total_minutos = (hora_saida - hora_entrada) * 60 + (minuto_saida - minuto_entrada)
  tempo_permanencia = (hora_saida - hora_entrada) * 60 + (minuto_saida - minuto_entrada)
  horas = total_minutos // 60
  minutos = total_minutos % 60
  total_a_pagar = horas * 7.00
  return horas, minutos, total_a_pagar

```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_hora(mensagem):
    while True:
        try:
            hora = int(input(mensagem))
            if hora < 0 or hora > 23:
                print("Digite uma hora válida (0 a 23)")
            else:
                return hora
        except ValueError:
            print("Digite um valor válido")


def ler_minuto(mensagem):
    while True:
        try:
            minuto = int(input(mensagem))
            if minuto < 0 or minuto > 59:
                print("Digite um minuto válido (0 a 59)")
            else:
                return minuto
        except ValueError:
            print("Digite um valor válido")
```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
hora_entrada = ler_hora("Hora de entrada: ")
minuto_entrada = ler_minuto("Minuto de entrada: ")
hora_saida = ler_hora("Hora de saída: ")
minuto_saida = ler_minuto("Minuto de saída: ")

horas, minutos, total_a_pagar = diferenca_estacionamento(hora_entrada, minuto_entrada, hora_saida, minuto_saida)

print(f"Tempo de permanência: {horas:02d}:{minutos:02d}")
print(f"Total a pagar: R$ {total_a_pagar:.2f}")
```

# EXERCÍCIO 8: EXERCÍCIO SORTEADO NÚMERO 16
Faça um programa Python modularizado que:
(i) Tenha uma função que calcule e retorne a área de um círculo de raio R;
(ii) Tenha outra função que calcule e retorne o volume do cilindro reto.
A função de (ii) deve usar a função de (i).
O raio e a altura de um cilindro devem ser lidos do teclado na célula do programa principal.
Obs.: Área do círculo: A = π . r2
Volume do Cilindro: Área do Círculo da Base vezes a Altura.
image.png

```
# CRIAR A FUNÇÃO AQUI
import math
def area_circulo (raio):
  return math.pi * raio**2

def volume_cilindro(raio, altura):
  area_base = area_circulo(raio)
  return area_base * altura
```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_positivo(mensagem):
    while True:
        try:
            valor = float(input(mensagem))
            if valor <= 0:
                print("O valor deve ser maior que zero.")
            else:
                return valor
        except ValueError:
            print("Digite um valor válido")
```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
raio = ler_positivo("Digite o raio do cilindro:")
altura = ler_positivo("DIgite a altura do cilindro:")

volume = volume_cilindro(raio, altura)

print(f"Área da base: {area_circulo(raio):.2f}")
print(f"Volume do cilindro: {volume:.2f}")
```

# EXERCÍCIO 9: EXERCÍCIO SORTEADO NÚMERO 17
Faça um programa Python modularizado que exiba o público total de um jogo de futebol e forneça a
arrecadação do jogo, sabendo que:
✓ Crianças abaixo de 10 anos não pagam;
✓ Jovens de 11 a 17 pagam 1⁄2 entrada;
✓ Acima dos 18 anos paga 1⁄2 entrada se doarem um quilo de alimento não perecível.
✓ O valor inteiro do ingresso é lido do usuário em reais (R$).
Na célula do programa principal, use a função várias vezes criando um Menu para o usuário.

```
# CRIAR A FUNÇÃO AQUI
def valor_pago(idade, valor, doacao):
  if idade < 10:
    return 0
  elif idade <= 18:
    return valor / 2
  elif idade >= 18 and doacao:
    return valor / 2
  elif idade >= 18:
    return valor
```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_idade():
  while True:
    try:
      idade = int(input("Digite a sua idade:"))
      if idade < 0:
        print("A idade deve ser maior que zero.")
      else:
          return idade
    except ValueError:
      print("Digite um valor válido")

def ler_valor():
  while True:
    try:
      valor = float(input("Digite o valor do ingresso (R$):"))
      if valor <= 0:
        print("O valor deve ser maior que zero.")
      else:
        return valor
    except ValueError:
      print("Digite um valor válido")

def ler_doacao():
    while True:
        resposta = input("Doou 1 kg de alimento não perecível? (s/n): ").lower()
        if resposta == "s":
            return True
        elif resposta == "n":
            return False
        else:
            print("Digite s ou n.")
```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
valor_inteiro = ler_valor()
publico = 0
arrecadacao = 0.0

while True:
    print("\nMENU")
    print("1: Registrar espectador")
    print("2: Mostrar público e arrecadação")
    print("0: Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        idade = ler_idade()
        doou = ler_doacao() if idade >= 18 else False
        arrecadacao += valor_pago(idade, valor_inteiro, doou)
        publico += 1
    elif opcao == "2":
        print(f"Público total: {publico}")
        print(f"Arrecadação: R$ {arrecadacao:.2f}")
    elif opcao == "0":
        print("Encerrando o programa.")
        break
    else:
        print("Opção inválida")
```

# EXERCÍCIO 10: EXERCÍCIO SORTEADO NÚMERO 19
Faça uma função em Python que receba três valores A, B, C e verifique se os mesmos podem formar
um triângulo ou não. Caso possam, dizer que tipo: Escaleno, Isóscele ou Eqüilátero (0, 2 ou 3 lados
iguais, respectivamente).
Condição obrigatória: É triângulo se, somente se (A < B + C) e (B < A + C) e (C < A + B).
Na célula do programa principal, use a função várias vezes, criando um Menu para o usuário.

```
# CRIAR A FUNÇÃO AQUI
def validar_triangulo(a, b, c):
    if not ((a < b + c) and (b < a + c) and (c < a + b)):
        return "Não forma um triângulo"

    if a == b and b == c:
        return "Triângulo Equilátero (3 lados iguais)"
    elif a == b or a == c or b == c:
        return "Triângulo Isósceles (2 lados iguais)"
    else:
        return "Triângulo Escaleno (0 lados iguais)"
```

```
# MODULARIZAÇÃO LEITURA / TRATAMENTO DE ERRO: Dados de entrada
def ler_lado(nome):
    while True:
        try:
            valor = float(input(f"Digite o valor de {nome}: "))
            if valor <= 0:
                print("O valor deve ser maior que zero.")
            else:
                return valor
        except ValueError:
            print("Digite um valor válido")
```

```
# INTERFACE (I/O) COM MODULARIZAÇÃO: Dados de entrada
while True:
    print("\nMENU")
    print("1: Verificar triângulo")
    print("0: Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        a = ler_lado("A")
        b = ler_lado("B")
        c = ler_lado("C")
        print(validar_triangulo(a, b, c))
    elif opcao == "0":
        print("Encerrando o programa.")
        break
    else:
        print("Opção inválida")
```
