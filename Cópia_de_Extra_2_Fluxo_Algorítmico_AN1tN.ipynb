```
# ALGORITMO: ENTRADA (Usuário) + PROCESSAMENTO (ULA) + SAÍDA (Objetivo)

try:
    # ENTRADA (INPUT)
    pontoPx = float(input('Digite a coordenada do ponto Px: '))
    pontoPy = float(input('Digite a coordenada do ponto Py: '))
    pontoQx = float(input('Digite a coordenada do ponto Qx: '))
    pontoQy = float(input('Digite a coordenada do ponto Qy: '))

    # PROCESSAMENTO (ULA)
    distancia = (((pontoQx - pontoPx) ** 2) + ((pontoQy - pontoPy) ** 2)) ** 0.5

    # SAÍDA (PRINT)
    print(f'O valor da distância é: {distancia:.2f}')
```

# Exercício 2

```
try:
  #ENTRADA (INPUT)
  ganhoHora = float(input("Digite o ganho por hora: "))
  quantiaHora = float(input('Digite quantas horas você trabalha:'))

  #CALCULOS
  salarioBruto = ganhoHora * quantiaHora
  impostoRenda = salarioBruto * 0.11
  inss = salarioBruto * 0.08
  sindicato =  salarioBruto * 0.05
  descontos = impostoRenda + inss + sindicato
  salarioLiquido = salarioBruto - descontos

  #SAÍDA (PRINT)
  print(f'Seu salário bruto é:', salarioBruto)
  print(f'Você pagou ao INSS: {inss: .2f}')
  print(f'Você pagou ao Sindicato: {sindicato: .2f}')
  print(f'Você teve de salário líquido: {salarioLiquido: .2f}')

except ValueError:
  print('Digite um valor válido')
```

# Exercício 3

Escrever um algoritmo em Python que leia o Tempo (float) em horas e exiba na tela, no formato:

*   HORA - HH
*   MINUTOS - MM
*  SEGUNDOS - SS

EXEMPLO:

ENTRADA: Tempo = 10.87 Horas

SAÍDA: Tempo = HH:MM:SS - 10:?:?

```
try:
  #ENTRADA (INPUT)
  hh = input(float('Insira a quantidade de horas:'))
  mm = input(float('Insira a quantidade de minutos:'))
  ss = input(float('Insira a quantidade de segundos:'))

  #PROCESSAMENTO
  tempo = hh, mm, ss

  #SAÍDA (PRINT)
  print(f'Tempo: {tempo}')

except ValueError:
  print('Digite um valor válido')
```

```
try:
  #ENTRADA (INPUT)
  tempo = float(input('Digite o tempo em horas:'))

  #PROCESSAMENTO
  hh = int(tempo)
  minutos = (tempo - hh) * 60
  mm = int(minutos)
  segundos = int((minutos - mm) * 60)

  #SAÍDA (PRINT)
  print(f'O tempo em horas é: {hh}: {mm}: {segundos}')

except ValueError:
    print('Digite um valor válido')
```
