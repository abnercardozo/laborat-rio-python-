2) Escreva um algoritmo em Python que exiba todos os números múltiplos de 6 e não múltiplos de 4 compreendidos no intervalo [100, 300], além da soma e da quantidade desses números.

Resolva o mesmo problema de três formas diferentes:

a) Usando for com range(start, stop)
b) Usando while com condição de parada (while <condição>)
c) Usando while True com break

```python
soma = 0
quantidade = 0
for i in range (100, 300):
  if i % 6 == 0 and i % 4 != 0:
    soma += i
    quantidade += 1
    print(i)

    print(f"A soma dos números é {soma}")
    print(f"A quantidade de números é {quantidade}")
```

```python
soma = 0
quantidade = 0
i = 100

while i <= 300:

    if i % 6 == 0 and i % 4 != 0:
        soma += i
        quantidade += 1
        print(i)

    i += 1

print(f"A soma dos números é {soma}")
print(f"A quantidade de números é {quantidade}")
```

```python
#WHILE TRUE + BREAK
soma = 0
quantidade = 0
i = 100

while True:
    if i > 300:
        break
    if i % 6 == 0 and i % 4 != 0:
        soma += i
        quantidade += 1
        print(i)
        i += 1

print(f"A soma dos números é {soma}")
print(f"A quantidade dos números é {quantidade}")
```

Questão 3 (Função + Modularização + Menu)

Crie uma função e seu protótipo que receba o salário bruto de um funcionário e o número de dependentes, e retorne o salário líquido e o total de descontos, conforme as regras:

Descontos:

INSS: 11% do salário bruto
Plano de Saúde: 2% do salário bruto

Benefícios:

Vale Alimentação: Salário Mínimo (R$ 998,05) + 5% do salário bruto por dependente

Na célula do programa principal, crie um menu com as opções:

1: Calcular contracheque de um funcionário
0: Sair do program

```python
#MODULARIZACAO
def calcular_descontos(salario_bruto):
    inss = salario_bruto * 0.11
    plano_saude = salario_bruto * 0.02
    return inss + plano_saude

def calcular_beneficios(salario_minimo, salario_bruto, quantidade_dependentes):
    vale_alimentacao = salario_minimo + (salario_bruto* 0.05)
    return vale_alimentacao * quantidade_dependentes
```

```python
#LEITURA DOS DADOS
def ler_dados():
    salario_bruto = int(input("Informe seu salário:"))
    quantidade_dependentes = int(input("Informe a quantidade de dependentes"))
    return salario_bruto, quantidade_dependentes
```

```python
#MENU
salario_minimo = 998.05
print("====MENU====")
print("1: Para calcular o salário líquido")
print("0: Sair do programa")
opcao = int(input("Informe a opção desejada:"))

while opcao != 0:
    if opcao == 1:
        salario_bruto, quantidade_dependentes = ler_dados()
        descontos = calcular_descontos(salario_bruto)
        beneficios = calcular_beneficios(salario_minimo,salario_bruto, quantidade_dependentes)
        salario_liquido = salario_bruto - descontos + beneficios
        print(f"O salário bruto é {salario_bruto}")
        print(f"O salário líquido é de {salario_liquido:.2f}")
        print(f"O total de descontos é de {descontos:.2f}")
        print(f"O total de benefícios é de {beneficios:.2f}")
    elif opcao == 0:
        print("Saindo do programa")
        break
    else:
        print("Opção inválida!")

```

Questão 4 (Função + Modularização + Menu)

Escreva um algoritmo modularizado em Python que leia o preço de uma mercadoria e o percentual de reajuste (X%), e exiba o preço reajustado. O usuário deve escolher entre:

1: Acréscimo (soma o reajuste ao preço)
2: Desconto (subtrai o reajuste do preço)

Na célula do programa principal, crie um menu com as opções:

1: Reajustar preço de uma mercadoria
0: Sair do programa

```python
#MODULARIZACAO
def calcular_acrescimo(preco, reajuste):
    return preco + (preco * reajuste / 100)

def calcular_desconto(preco, reajuste):
    return preco - (preco * reajuste / 100)
```

```python
#LER DADOS
def ler_dados():
    preco = float(input("Informe o preço da sua mercadoria:"))
    reajuste = float(input("Informe o percentual de reajuste:"))
    return preco, reajuste
```

```python
while True:
    print("====MENU====")
    print("1: Para calcular o reajuste da mercadoria:")
    print("0: Sair do programa")
    opcao = int(input("Informe a opção desejada: "))

    if opcao == 1:
        escolha = int(input("Digite 1 para acréscimo e 2 para desconto: "))
        preco, reajuste = ler_dados()

        if escolha == 1:
            preco_reajustado = calcular_acrescimo(preco, reajuste)
            print(f"O preço reajustado é de R$ {preco_reajustado:.2f}")
        elif escolha == 2:
            preco_reajustado = calcular_desconto(preco, reajuste)
            print(f"O preço reajustado é de R$ {preco_reajustado:.2f}")
        else:
            print("Opção inválida")
    elif opcao == 0:
        print("Saindo do programa")
        break
    else:
        print("Opção inválida")
```
