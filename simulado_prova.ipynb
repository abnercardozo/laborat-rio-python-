2) Escreva um algoritmo em Python que exiba todos os números múltiplos de 6 e não múltiplos de 4 compreendidos no intervalo [100, 300], além da soma e da quantidade desses números.

Resolva o mesmo problema de três formas diferentes:

a) Usando for com range(start, stop)
b) Usando while com condição de parada (while <condição>)
c) Usando while True com break

```python
soma = 0
quantidade = 0
for i in range (100, 300):
  if i % 6 == 0 and i % 4 != 0:
    soma += i
    quantidade += 1
    print(i)

    print(f"A soma dos números é {soma}")
    print(f"A quantidade de números é {quantidade}")
```

```python
soma = 0
quantidade = 0
i = 100

while i <= 300:

    if i % 6 == 0 and i % 4 != 0:
        soma += i
        quantidade += 1
        print(i)

    i += 1

print(f"A soma dos números é {soma}")
print(f"A quantidade de números é {quantidade}")
```

```python
#WHILE TRUE + BREAK
soma = 0
quantidade = 0
i = 100

while True:
    if i > 300:
        break
    if i % 6 == 0 and i % 4 != 0:
        soma += i
        quantidade += 1
        print(i)
        i += 1

print(f"A soma dos números é {soma}")
print(f"A quantidade dos números é {quantidade}")
```

Questão 3 (Função + Modularização + Menu)

Crie uma função e seu protótipo que receba o salário bruto de um funcionário e o número de dependentes, e retorne o salário líquido e o total de descontos, conforme as regras:

Descontos:

INSS: 11% do salário bruto
Plano de Saúde: 2% do salário bruto

Benefícios:

Vale Alimentação: Salário Mínimo (R$ 998,05) + 5% do salário bruto por dependente

Na célula do programa principal, crie um menu com as opções:

1: Calcular contracheque de um funcionário
0: Sair do program

```python
#MODULARIZACAO
def calcular_descontos(salario_bruto):
    inss = salario_bruto * 0.11
    plano_saude = salario_bruto * 0.02
    return inss + plano_saude

def calcular_beneficios(salario_minimo, salario_bruto, quantidade_dependentes):
    vale_alimentacao = salario_minimo + (salario_bruto* 0.05)
    return vale_alimentacao * quantidade_dependentes
```

```python
#LEITURA DOS DADOS
def ler_dados():
    salario_bruto = int(input("Informe seu salário:"))
    quantidade_dependentes = int(input("Informe a quantidade de dependentes"))
    return salario_bruto, quantidade_dependentes
```

```python
#MENU
salario_minimo = 998.05
print("====MENU====")
print("1: Para calcular o salário líquido")
print("0: Sair do programa")
opcao = int(input("Informe a opção desejada:"))

while opcao != 0:
    if opcao == 1:
        salario_bruto, quantidade_dependentes = ler_dados()
        descontos = calcular_descontos(salario_bruto)
        beneficios = calcular_beneficios(salario_minimo,salario_bruto, quantidade_dependentes)
        salario_liquido = salario_bruto - descontos + beneficios
        print(f"O salário bruto é {salario_bruto}")
        print(f"O salário líquido é de {salario_liquido:.2f}")
        print(f"O total de descontos é de {descontos:.2f}")
        print(f"O total de benefícios é de {beneficios:.2f}")
    elif opcao == 0:
        print("Saindo do programa")
        break
    else:
        print("Opção inválida!")

```

Escreva um algoritmo modularizado em Python que leia o preço de uma mercadoria e o percentual de reajuste (X%), e exiba o preço reajustado. O usuário deve escolher entre:

1: Acréscimo (soma o reajuste ao preço)
2: Desconto (subtrai o reajuste do preço)

Na célula do programa principal, crie um menu com as opções:

1: Reajustar preço de uma mercadoria
0: Sair do programa

```python
#MODULARIZACAO
def calcular_acrescimo(preco, reajuste):
    return preco + (preco * reajuste / 100)

def calcular_desconto(preco, reajuste):
    return preco - (preco * reajuste / 100)
```

```python
#LER DADOS
def ler_dados():
    preco = float(input("Informe o preço da sua mercadoria:"))
    reajuste = float(input("Informe o percentual de reajuste:"))
    return preco, reajuste
```

```python
while True:
    print("====MENU====")
    print("1: Para calcular o reajuste da mercadoria:")
    print("0: Sair do programa")
    opcao = int(input("Informe a opção desejada: "))

    if opcao == 1:
        escolha = int(input("Digite 1 para acréscimo e 2 para desconto: "))
        preco, reajuste = ler_dados()

        if escolha == 1:
            preco_reajustado = calcular_acrescimo(preco, reajuste)
            print(f"O preço reajustado é de R$ {preco_reajustado:.2f}")
        elif escolha == 2:
            preco_reajustado = calcular_desconto(preco, reajuste)
            print(f"O preço reajustado é de R$ {preco_reajustado:.2f}")
        else:
            print("Opção inválida")
    elif opcao == 0:
        print("Saindo do programa")
        break
    else:
        print("Opção inválida")
```

Questão 3 (Função + Modularização)

Crie uma função e seu protótipo que receba um número inteiro qualquer positivo e retorne True se for PRIMO ou False caso contrário (exercício 5 da lista de modularização).

Regras: 1, 2 e 3 são primos; os demais só são primos se forem múltiplos apenas de 1 e de si mesmos.

Na célula do programa principal, crie um menu:

1: Verificar se um número é primo
0: Sair do programa

```python
#FUNCAO
def verificar_primo(numero):
    if numero <= 1:
        return False
    if numero <= 3:
        return True
    if numero % 2 == 0 or numero % 3 == 0:
        return False
    if numero % 1 == 0 and numero % numero == 0:
        return True
```

```python
#LER DADOS
def ler_numero():
  while True:
    numero = int(input("Informe um número acima de zero:"))
    if numero > 0:
      return numero
    else:
      print("Número inválido")
      break
```

```python
#MENU
print("====MENU====")
print("1: Para verificar se o número é primo")
print("0: Sair do programa")
print("Escolha sua opção")

opcao = int(input())
if opcao == 1:
  numero = ler_numero()
  if verificar_primo(numero):
    print(f"O número {numero} é primo")
  else:
    print(f"O número {numero} não é primo")
elif opcao == 0:
  print("Saindo do programa")
```

Questão 4 (Função + Modularização)

Crie duas funções e seus protótipos que calculem o Arranjo e a Combinação de n elementos tomados p a p (exercício 3 da lista de modularização):

$$Arranjo(n, p) = \frac{n!}{(n-p)!} \qquad Combinação(n, p) = \frac{Arranjo(n, p)}{p!}$$

Sabendo que n e p ≥ 0 e n ≥ p. A função de Combinação deve usar a função de Arranjo.

Na célula do programa principal, crie um menu:

1: Calcular Arranjo e Combinação
0: Sair do programa

Valide se n e p atendem às regras antes de calcular.

```python
import math
def arranjo(n,p):
  if n >= p:
    resultado = math.factorial(n) / math.factorial(n-p)
    return resultado
def combinacao(n,p):
  if n >= p:
    resultado = arranjo(n,p)/ math.factorial(p)
    return resultado

```

```python
#LER N E P
def ler_num():
  while True:
    n = int(input("Informe o valor de n:"))
    p = int(input("Informe o valor de p:"))
    if n >= 0 and p >= 0 and n >= p:
      return n, p
    else:
      print("Os valores devem atender: n >= 0, p >= 0 e n >= p")
```

```python
#MENU
print("====MENU====")
print("1: Para calcular o arranjo e a combinação")
print("0: Para sair do programa")
print("Digite sua opção")
opcao = int(input())

if opcao == 1:
  n, p = ler_num()
  resultado = arranjo(n,p)
  resultado1 = combinacao(n,p)
  print(f"O arranjo tem como resultado {resultado}")
  print(f"A combinação tem como resultado {resultado1}")
elif opcao == 0:
  print("Saindo do programa")
```

E retorne o resultado da operação. Trate o caso de divisão por zero.

Na célula do programa principal, crie um menu que repita a operação enquanto o usuário desejar.

```python
def operador(num1, num2, operador):

    if operador == 1:
        return num1 + num2

    elif operador == 2:
        return num1 - num2

    elif operador == 3:
        return num1 * num2

    elif operador == 4:
        if num2 == 0:
            return print("Número não pode ser dividido por zero")
        else:
            return num1 / num2

    elif operador == 5:
        if num2 == 0:
            return print("Número não pode ser dividido por zero")
        else:
            return num1 % num2

    elif operador == 6:
        if num2 == 0:
            return print("Número não pode ser dividido por zero")
        else:
            return num1 // num2
```

Questão 6 (Função + Modularização)

Crie uma função e seu protótipo que determine o Enésimo termo (An) e a Soma dos termos (Sn) de uma Progressão Geométrica (exercício 15 da lista de modularização):

$$An = a_1 \cdot q^{(n-1)} \qquad Sn = \frac{a_1 \cdot (q^n - 1)}{q - 1}$$

Na célula do programa principal, leia a1, q e n do usuário e exiba An e Sn.

```python
#MODULARIZACAO
def calculo_an(a1, q, n):
    return a1 * (q ** (n - 1))

def calculo_sn(a1, q, n):
    return (a1 * (q ** n - 1)) / (q - 1)
```

```python
#LER DADOS
def ler_num():
    a1 = int(input("Informe o valor de a1:"))
    q = int(input("Informe o valor de q:"))
    n = int(input("Informe o valor de n:"))
    return a1, q, n
```

```python
#MENU
while True:
  try:
    a1, q, n = ler_num()
    resultado = calculo_an(a1, q, n)
    resultado1 = calculo_sn(a1, q, n)
    print(f"O enésimo termo é {resultado}")
    print(f"A soma dos termos é {resultado1}")
    break
  except ValueError:
    print("Valor inválido")
```
