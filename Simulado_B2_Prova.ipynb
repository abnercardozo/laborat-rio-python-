## Letra C)

Exiba um relatório administrativo com o **Total da Compra**:

* SUBTOTAL = somatória de (PREÇO x QUANTIDADE) de todos os itens.
* IMPOSTO = 10% sobre o valor (PREÇO x QUANTIDADE) **apenas** dos itens **sem** isenção de imposto. Exiba o imposto separado por QUANTIDADE comprada (1, 2 e 3 unidades).
* TOTAL A PAGAR = SUBTOTAL + IMPOSTO.

````python
subtotal = sum([item[2] * item[3] for item in tabela])

imposto1 = sum([item[2] * item[3] * 0.10 for item in tabela if item[4] == 'NÃO' and item[3] == 1])
imposto2 = sum([item[2] * item[3] * 0.10 for item in tabela if item[4] == 'NÃO' and item[3] == 2])
imposto3 = sum([item[2] * item[3] * 0.10 for item in tabela if item[4] == 'NÃO' and item[3] == 3])


#declarando total do imposto
totalImposto = imposto1 + imposto2 + imposto3

totalPagar = subtotal - totalImposto
````

````python
#relatorio
print("RELATÓRIO ADMINISTRATIVO: TOTAL DA COMPRA")
print(f"Subtotal: {subtotal:.2f}R$")
print(f"Imposto de 10% para uma unidade: {imposto1:.2f}%")
print(f"Imposto de 10% para duas unidades: {imposto2:.2f}%")
print(f"Imposto de 10% para três unidades: {imposto3:.2f}%")
print(f"Total a pagar: {totalPagar:.2f}R$")
````
